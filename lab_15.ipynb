```python

import math

# Алхам 1. Хавтгайн цэгийг илэрхийлэх Vector класс
class Vector:
    def __init__(self, x, y, name):
        self.x = x
        self.y = y
        self.name = name

    def hewleh(self):
        print(self.name + '(' + str(self.x) + ',' + str(self.y) + ')')


# Алхам 2. Vector классаас удамшсан Vector3D класс
class Vector3D(Vector):
    def __init__(self, x, y, z, name):
        super().__init__(x, y, name)
        self.z = z

    def hewleh(self):
        print(self.name + '(' + str(self.x) + ',' +
              str(self.y) + ',' + str(self.z) + ')')

    # Алхам 4. Хоёр цэгийн хоорондох зай
    def zai(self, other):
        zz = math.sqrt((self.x - other.x)**2 +
                       (self.y - other.y)**2 +
                       (self.z - other.z)**2)
        print('{} ба {} 2 цэгийн хоорондох зай {}'.format(
            self.name, other.name, zz))

    # Алхам 6. Зайг хэвлэх метод
    def zaihewleh(self, other):
        zz = math.sqrt((self.x - other.x)**2 +
                       (self.y - other.y)**2 +
                       (self.z - other.z)**2)
        print('{} ба {} 2 цэгийн хоорондох зай {}'.format(
            self.name, other.name, zz))

    # Зайг тооцоолж утга буцаах метод
    def zaiutga(self, other):
        zz = math.sqrt((self.x - other.x)**2 +
                       (self.y - other.y)**2 +
                       (self.z - other.z)**2)
        return zz


# Алхам 3. Хоёр объект үүсгэх
v1 = Vector3D(1, 2, -1, 'A')
v2 = Vector3D(5, -2, -3, 'B')

v1.hewleh()
v2.hewleh()

# Алхам 5. Хоёр цэгийн зайг олох
v1.zai(v2)

# Алхам 6. Дөрвөн цэг үүсгэх
v3 = Vector3D(0, 0, 10, 'C')
v4 = Vector3D(8, -2, 0, 'D')

v3.hewleh()
v4.hewleh()

# Бүх боломжит хосууд
lines = [(v1, v2), (v1, v3), (v1, v4),
         (v2, v3), (v2, v4), (v3, v4)]

maxzai = 0
maxind = 0

for i in range(len(lines)):
    zai = lines[i][0].zaiutga(lines[i][1])
    lines[i][0].zaihewleh(lines[i][1])

    if zai > maxzai:
        maxind = i
        maxzai = zai

print('------ Хамгийн хол цэгүүд ------')
lines[maxind][0].zaihewleh(lines[maxind][1])
```

```python

# Тоогоор үржих үйлдлийг Vector3D классд нэмэх
def urjih(self, num):
    v = Vector3D(0, 0, 0, 'C')
    v.x = self.x * num
    v.y = self.y * num
    v.z = self.z * num
    return v

Vector3D.__mul__ = urjih

# Объект үүсгэх
v1 = Vector3D(1, 2, -1, 'A')

v1.hewleh()

print('-----------v1 векторыг 3-аар үржихэд---------------')

v2 = v1 * 3
v2.hewleh()
```

```python

class bankAccount:
    def __init__(self):
        self.ner = ""
        self.dugaar = ""
        self.uldegdel = 0
        self.neelttei = False

    def neeh(self):
        if self.neelttei:
            print("Данс аль хэдийн нээгдсэн.")
            return
        self.ner = input("Харилцагчийн нэр: ")
        self.dugaar = input("Дансны дугаар: ")
        self.uldegdel = 0
        self.neelttei = True
        print("Данс амжилттай нээгдлээ.")

    def mongo_hiih(self, mongo):
        if not self.neelttei:
            print("Данс хаалттай байна.")
        elif mongo <= 0:
            print("Мөнгөний дүн эерэг байх ёстой.")
        else:
            self.uldegdel += mongo
            print("Мөнгө нэмэгдлээ.")

    def mongo_avah(self, mongo):
        if not self.neelttei:
            print("Данс хаалттай байна.")
        elif mongo <= 0:
            print("Мөнгөний дүн эерэг байх ёстой.")
        elif mongo > self.uldegdel:
            print("Үлдэгдэл хүрэлцэхгүй.")
        else:
            self.uldegdel -= mongo
            print("Мөнгө авлаа.")

    def uldegdel_harah(self):
        if self.neelttei:
            print("Үлдэгдэл:", self.uldegdel)
        else:
            print("Данс хаалттай.")

    def haah(self):
        if not self.neelttei:
            print("Данс аль хэдийн хаалттай.")
        elif self.uldegdel != 0:
            print("Данс хаахын өмнө үлдэгдлээ 0 болгоно уу.")
        else:
            self.neelttei = False
            print("Данс хаагдлаа.")


b = bankAccount()
b.neeh()
b.mongo_hiih(100000)
b.uldegdel_harah()
b.mongo_avah(30000)
b.uldegdel_harah()
b.mongo_avah(70000)
b.haah()
```

```python

class time:
    def __init__(self, hours=0, minutes=0, seconds=0):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds

    def orolt(self):
        self.hours = int(input("Цаг: "))
        self.minutes = int(input("Минут: "))
        self.seconds = int(input("Секунд: "))

        if not (0 <= self.hours < 24 and
                0 <= self.minutes < 60 and
                0 <= self.seconds < 60):
            raise ValueError("Цагийн утга буруу байна.")

    def garalt(self):
        print(f"{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}")

    def niit_second(self):
        return self.hours * 3600 + self.minutes * 60 + self.seconds

    def hoorondoh_zai(self, other):
        z = abs(self.niit_second() - other.niit_second())
        h = z // 3600
        m = (z % 3600) // 60
        s = z % 60
        print(f"Хоёр цагийн зөрүү: {h:02d}:{m:02d}:{s:02d}")


t1 = time()
t2 = time()

print("Эхний цаг:")
t1.orolt()

print("Хоёр дахь цаг:")
t2.orolt()

t1.garalt()
t2.garalt()
t1.hoorondoh_zai(t2)
```

```python

class Complex:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def __truediv__(self, other):
        d = other.a**2 + other.b**2

        if d == 0:
            raise ZeroDivisionError("Тэг комплекс тоонд хувааж болохгүй")

        x = (self.a * other.a + self.b * other.b) / d
        y = (self.b * other.a - self.a * other.b) / d

        return Complex(x, y)

    def hewleh(self):
        print(f"{self.a:g}{self.b:+g}i")

a = int(input("Эхний тооны бодит хэсэг: "))
b = int(input("Эхний тооны хуурмаг хэсэг: "))
c = int(input("Хоёр дахь тооны бодит хэсэг: "))
d = int(input("Хоёр дахь тооны хуурмаг хэсэг: "))

c1 = Complex(a, b)
c2 = Complex(c, d)

print("Хуваалтын үр дүн:")
c3 = c1 / c2
c3.hewleh()
```

```python

class Complex:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def __mul__(self, other):
        x = self.a * other.a - self.b * other.b
        y = self.a * other.b + self.b * other.a
        return Complex(x, y)

    def hewleh(self):
        print(f"{self.a}{self.b:+}i")

a = int(input("Эхний тооны бодит хэсэг: "))
b = int(input("Эхний тооны хуурмаг хэсэг: "))
c = int(input("Хоёр дахь тооны бодит хэсэг: "))
d = int(input("Хоёр дахь тооны хуурмаг хэсэг: "))

c1 = Complex(a, b)
c2 = Complex(c, d)

print("Үржвэр:")
c3 = c1 * c2
c3.hewleh()
```

```python

# Энгийн удамшил

class Student:
    def __init__(self, ner, nas):
        self.ner = ner
        self.nas = nas

    def medeelel(self):
        print("Нэр:", self.ner)
        print("Нас:", self.nas)


class UniversityStudent(Student):
    def __init__(self, ner, nas, mergejil):
        super().__init__(ner, nas)
        self.mergejil = mergejil

    def garalt(self):
        self.medeelel()
        print("Мэргэжил:", self.mergejil)


ner = input("Оюутны нэр: ")
nas = int(input("Нас: "))
mergejil = input("Мэргэжил: ")

s = UniversityStudent(ner, nas, mergejil)
print("----- Оюутны мэдээлэл -----")
s.garalt()
```
